```python
import pandas as pd

# Load the dataset
df = pd.read_csv("data.csv")

# Define education level mapping to years (points)
education_points = {
    "Master's": 17,
    "Bachelor's": 15,
    "Doctorate": 20,
    "High School": 12,
    "Ingénieur": 17,
    "No Formal Education": 0,
    "Technical Diploma": 14
}

# Create the new column based on a column named "education"
df["education_years"] = df["education_level"].map(education_points)

# Save it back if you want
df.to_csv("data_with_education_years.csv", index=False)
```

```python
import pandas as pd

# Load the second dataset
df = pd.read_csv("emploitic_enhanced_job_listings_final.csv")

# Function to infer job type
def infer_job_type(row):
    text = f"{row.get('job title', '')} {row.get('description', '')}".lower()
    if 'temps partiel' in text or 'part-time' in text or 'temps part time' in text:
        return 'Part-time'
    elif 'temps plein' in text or 'full-time' in text:
        return 'Full-time'
    else:
        if 'freelance' in text or 'indépendant' in text:
            return 'Part-time'
        return 'Full-time'  # Default fallback

# Apply inference
df['job_type'] = df.apply(infer_job_type, axis=1)

# Save the result
df.to_csv("emploitic_enhanced_with_job_type.csv", index=False)

print("✅ job_type column added and saved as emploitic_enhanced_with_job_type.csv")
```
