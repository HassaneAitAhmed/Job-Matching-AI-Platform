```python
import pandas as pd

# Define a mapping of education levels to approximate years of education

education_years_mapping = {
    "High School": 9,
    "No Formal Education": 0,
    "Bachelor's": 12,
    "Master's": 17,
    "Doctorate": 20,
    "Ingénieur": 17
}

# Load the CSV file
input_csv = "Temp.csv"  # Replace with your input file name
output_csv = "data.csv"  # Desired output file name

# Read the CSV file into a DataFrame
df = pd.read_csv(input_csv)

# Ensure the column containing education levels is named 'Education Level' or adjust accordingly
if 'education_level' in df.columns:
    # Map the education levels to approximate years of education
    df['Approximate Years of Education'] = df['education_level'].map(education_years_mapping)

    # Save the updated DataFrame to a new CSV file
    df.to_csv(output_csv, index=False)
    print(f"Converted CSV saved to {output_csv}")
else:
    print("The column 'Education Level' was not found in the input CSV file.")
```
