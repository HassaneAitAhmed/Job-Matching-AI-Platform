```python
import pandas as pd
import random
import numpy as np

# ---------- 1. Load Data ----------
# Employees (seekers) are loaded from a dataset CSV
seekers_df = pd.read_csv('emplo.csv')    # Columns: seeker_id, skills, education_level, experience_years, desired_salary, location, sector_interest, contract_type_interest

# Job posts are entered interactively by the user
def get_job_posts_from_user():
    posts = []
    n = int(input("Enter number of job posts: "))
    for i in range(n):
        print(f"--- Job {i+1} ---")
        job_id          = input("Job ID: ")
        required_skills = input("Required skills (comma-separated): ")
        min_education   = input("Minimum education (e.g. Baccalaureate, License, Master, Engineer, Doctorate): ")
        min_experience  = int(input("Minimum years of experience: "))
        salary_offer    = float(input("Salaryoffer: "))
        location        = input("Location: ")
        sector          = input("Sector: ")
        contract_type   = input("Contract type: ")
        posts.append({
            'job_id': job_id,
            'required_skills': required_skills,
            'min_education': min_education,
            'min_experience': min_experience,
            'salary_offer': salary_offer,
            'location': location,
            'sector': sector,
            'contract_type': contract_type
        })
    return pd.DataFrame(posts)

jobs_df = get_job_posts_from_user()

# ---------- 2. Validation ----------
required_seeker_cols = {'seeker_id','skills','education_level','experience_years','desired_salary','location','sector_interest','contract_type_interest'}
required_job_cols     = {'job_id','required_skills','min_education','min_experience','salary_offer','location','sector','contract_type'}
assert required_seeker_cols.issubset(seekers_df.columns), f"Missing seeker cols: {required_seeker_cols - set(seekers_df.columns)}"
assert required_job_cols.issubset(jobs_df.columns), f"Missing job cols: {required_job_cols - set(jobs_df.columns)}"

# ---------- 3. Preprocessing Helpers ----------
def parse_skills(s):
    return set(skill.strip().lower() for skill in str(s).split(',') if skill.strip())

# Education ranking
EDU_RANK = {'Baccalaureate': 1, 'License': 2, 'Master': 3, 'Engineer': 4, 'Doctorate': 5}

# Build skill sets and rank fields
seekers_df['skill_set'] = seekers_df['skills'].apply(parse_skills)
jobs_df   ['skill_set'] = jobs_df   ['required_skills'].apply(parse_skills)
seekers_df['edu_rank']  = seekers_df['education_level'].map(lambda e: EDU_RANK.get(e, 0))
jobs_df   ['edu_rank']  = jobs_df   ['min_education'].map(lambda e: EDU_RANK.get(e, 0))

# Precompute normalization constants
global_max_exp   = seekers_df['experience_years'].max() or 1
max_diff_salary  = seekers_df['desired_salary'].max() + jobs_df['salary_offer'].max()

# Problem size constants
N_seekers = len(seekers_df)
N_jobs    = len(jobs_df)
NO_MATCH  = -1

# ---------- 4. GA Parameters ----------
WEIGHTS = {
    'skills':        0.2,
    'experience':    0.2,
    'salary':        0.15,
    'sector':        0.15,
    'education':     0.15,
    'location':      0.15,
    'contract_type': 0.2
}
# Normalize weights to sum=1
total_w = sum(WEIGHTS.values())
assert total_w > 0, "Total weight must be positive."
for k in WEIGHTS:
    WEIGHTS[k] /= total_w

# ---------- 5. GA Operators ----------
# Selection
def tournament_selection(pop, scores, k=3):
    selected = []
    for _ in pop:
        tour = random.sample(list(zip(pop, scores)), k)
        winner = max(tour, key=lambda x: x[1])[0]
        selected.append(winner)
    return selected

def roulette_wheel_selection(pop, scores):
    positive = [max(0, s) for s in scores]
    total = sum(positive)
    if total <= 0:
        return random.choices(pop, k=len(pop))
    probs = [s/total for s in positive]
    return random.choices(pop, weights=probs, k=len(pop))

# Crossover
def onepoint_crossover(p1, p2):
    cut = random.randint(1, N_seekers-1)
    return p1[:cut] + p2[cut:], p2[:cut] + p1[cut:]

def pmx_crossover(p1, p2):
    a, b = sorted(random.sample(range(N_seekers), 2))
    def pmx(parent1, parent2):
        child = [None] * N_seekers
        child[a:b] = parent1[a:b]
        for i in list(range(a)) + list(range(b, N_seekers)):
            gene = parent2[i]
            while gene in child[a:b]:
                idx = parent2.index(parent1[parent2.index(gene)])
                gene = parent2[idx]
            child[i] = gene
        return child
    return pmx(p1, p2), pmx(p2, p1)

def order_crossover(p1, p2):
    a, b = sorted(random.sample(range(N_seekers), 2))
    def ox(parent1, parent2):
        child = [None] * N_seekers
        child[a:b] = parent1[a:b]
        fill = [g for g in parent2 if g not in child]
        idx = b
        for gene in fill:
            child[idx % N_seekers] = gene
            idx += 1
        return child
    return ox(p1, p2), ox(p2, p1)

# Mutation
def mutate(chrom, rate=0.05):
    for i in range(N_seekers):
        if random.random() < rate:
            valid = [j for j in range(N_jobs)
                     if seekers_df.iloc[i].edu_rank >= jobs_df.iloc[j].edu_rank
                     and seekers_df.iloc[i].experience_years >= jobs_df.iloc[j].min_experience]
            chrom[i] = random.choice(valid) if valid else NO_MATCH
    return chrom

# ---------- 6. Fitness Function ----------
def fitness(chrom):
    total_score = 0
    for i, job_idx in enumerate(chrom):
        seeker = seekers_df.iloc[i]
        if job_idx == NO_MATCH:
            continue
        job = jobs_df.iloc[job_idx]
        # Hard constraints
        if seeker.edu_rank < job.edu_rank or seeker.experience_years < job.min_experience:
            return -1e6
        # Compute sub-scores
        s, j = seeker.skill_set, job.skill_set
        skill_score     = len(s & j) / len(s | j) if (s | j) else 0
        exp_score       = seeker.experience_years / global_max_exp
        salary_score    = 1 - (abs(job.salary_offer - seeker.desired_salary) / max_diff_salary)
        sector_score    = int(seeker['sector_interest'] == job['sector'])
        education_score = seeker.edu_rank / max(EDU_RANK.values())
        loc_score       = int(seeker['location'] == job['location'])
        contract_score  = int(seeker['contract_type_interest'] == job['contract_type'])
        # Weighted sum
        score = (
            WEIGHTS['skills'] * skill_score +
            WEIGHTS['experience'] * exp_score +
            WEIGHTS['salary'] * salary_score +
            WEIGHTS['sector'] * sector_score +
            WEIGHTS['education'] * education_score +
            WEIGHTS['location'] * loc_score +
            WEIGHTS['contract_type'] * contract_score
        )
        total_score += score
    return total_score

# ---------- 7. GA Runner ----------
def run_ga(selection='tournament', crossover='onepoint', pop_size=50, gens=100, mut_rate=0.02, elite=2):
    # Initialize population
    pop = []
    for _ in range(pop_size):
        chrom = []
        for i in range(N_seekers):
            valid = [j for j in range(N_jobs)
                     if seekers_df.iloc[i].edu_rank >= jobs_df.iloc[j].edu_rank
                     and seekers_df.iloc[i].experience_years >= jobs_df.iloc[j].min_experience]
            chrom.append(random.choice(valid) if valid and random.random() < 0.9 else NO_MATCH)
        pop.append(chrom)
    best, best_score = None, -np.inf
    for _ in range(gens):
        scores = [fitness(c) for c in pop]
        idx = int(np.argmax(scores))
        if scores[idx] > best_score:
            best_score, best = scores[idx], pop[idx].copy()
        # Selection
        mating = tournament_selection(pop, scores) if selection=='tournament' else roulette_wheel_selection(pop, scores)
        # Elitism
        ranked = sorted(zip(pop, scores), key=lambda x: x[1], reverse=True)
        new_pop = [c for c, _ in ranked[:elite]]
        # Offspring
        while len(new_pop) < pop_size:
            p1, p2 = random.sample(mating, 2)
            if crossover=='onepoint': c1,c2 = onepoint_crossover(p1,p2)
            elif crossover=='pmx':    c1,c2 = pmx_crossover(p1,p2)
            else:                     c1,c2 = order_crossover(p1,p2)
            new_pop.extend([mutate(c1, mut_rate), mutate(c2, mut_rate)])
        pop = new_pop[:pop_size]
    matched = sum(g != NO_MATCH for g in best)
    success_rate = matched / N_seekers
    margin_error = 1.96 * np.sqrt(success_rate*(1-success_rate)/N_seekers)
    return {'best':best, 'fitness':best_score, 'success_rate':success_rate, 'margin_error':margin_error}

# ---------- 8. Execute GA Variants ----------
if __name__ == '__main__':
    variants = [
        ('tournament','onepoint'),
        ('roulette','onepoint'),
        ('tournament','pmx'),
        ('tournament','order')
    ]
    for sel, cross in variants:
        result = run_ga(selection=sel, crossover=cross)
        print(f"=== {sel}_{cross} ===")
        print(f"Fitness: {result['fitness']}")
        print(f"Success rate: {result['success_rate']:.2%}")
        print(f"Margin error: ±{result['margin_error']:.2%}\n")

# ---------- 9. Example Test Run ----------
# To test without interactive input, you can bypass get_job_posts_from_user() by
# loading sample job posts directly, e.g.:
#
seekers_df = pd.DataFrame([
    {'seeker_id':1,'skills':'python, sql','education_level':'Master','experience_years':5,'desired_salary':70000,'location':'Algiers','sector_interest':'IT','contract_type_interest':'Full-time'},
    {'seeker_id':2,'skills':'excel, finance','education_level':'License','experience_years':3,'desired_salary':50000,'location':'Oran','sector_interest':'Finance','contract_type_interest':'Part-time'}
 ])
 jobs_df = pd.DataFrame([
    {'job_id':'A','required_skills':'python, spark','min_education':'License','min_experience':2,'salary_offer':65000,'location':'Algiers','sector':'IT','contract_type':'Full-time'},
    {'job_id':'B','required_skills':'excel','min_education':'Baccalaureate','min_experience':1,'salary_offer':45000,'location':'Oran','sector':'Finance','contract_type':'Part-time'}
])
# Then rerun preprocessing steps 3 and execute GA as above to see results.
```
