```python
import pandas as pd

# Load the dataset
df = pd.read_csv("data_with_education_years.csv")

# Define a function to infer job type from job title or description
def infer_job_type(row):
    text = f"{row.get('job title', '')} {row.get('description', '')}".lower()
    if 'temps partiel' in text or 'part-time' in text or 'temps part time' in text:
        return 'Part-time'
    elif 'temps plein' in text or 'full-time' in text or 'temps plein' in text:
        return 'Full-time'
    else:
        # Fallback: guess based on common patterns
        if 'freelance' in text or 'indépendant' in text:
            return 'Part-time'
        return 'Full-time'  # Default assumption

# Apply function to create the job_type column
df['job_type'] = df.apply(infer_job_type, axis=1)

# Save updated file
df.to_csv("data_with_job_type.csv", index=False)

print("✅ job_type column added and file saved as data_with_job_type.csv")
```

```python
# Install required packages
!pip install faker pandas numpy

import pandas as pd
import numpy as np
from faker import Faker
from random import choices, randint, sample, choice, uniform, random
from datetime import datetime, timedelta

# Initialize Faker with Arabic locale for names
fake = Faker('ar_AA')
fake_other = Faker(['fr_FR'])  # For other data

# Algerian-specific configurations
ALGERIAN_CITIES = [
    'Algiers', 'Oran', 'Constantine', 'Annaba', 'Blida', 'Batna', 'Sétif',
    'Tlemcen', 'Béjaïa', 'Tiaret', 'Biskra', 'Tizi Ouzou', 'Mostaganem', 'Ghardaïa',
    'Chlef', 'El Oued', 'Skikda', 'Saïda', 'Laghouat', 'M\'sila', 'Jijel', 'Adrar',
    'Tébessa', 'Tindouf', 'Ouargla', 'Béchar', 'Tamanrasset', 'El Tarf', 'Illizi',
    'Naama', 'Aïn Témouchent', 'Timimoun', 'Djanet', 'In Salah'
]

ALGERIAN_SECTORS = [
    'Energy & Petroleum', 'Telecommunications', 'Banking & Finance',
    'Construction', 'Agriculture', 'Pharmaceuticals', 'Transport & Logistics',
    'Information Technology', 'Public Sector', 'Education', 'Healthcare',
    'Tourism & Hospitality', 'Manufacturing', 'Retail', 'Renewable Energy',
    'Aerospace & Defense', 'Mining', 'Fisheries', 'E-Commerce', 'Media & Entertainment'
]

ALGERIAN_COMPANIES = [
    'Sonatrach', 'Mobilis', 'BNA', 'CEVITAL', 'Condor Electronics',
    'Sofiane Pharma', 'Cosider', 'ENIE', 'Air Algérie', 'SNVI', 'Naftal',
    'Djezzy', 'Ooredoo Algeria', 'Cevital Agro-Industry', 'Sonalgaz',
    'Algérie Télécom', 'CNAN', 'Manadjim El Djazaïr', 'Batimétal'
] + [fake_other.company() + choice([' SPA', ' SARL', ' EURL']) for _ in range(50)]

# Expanded departments (30 departments)
DEPARTMENTS = {
    'Energy & Petroleum': [
        'Exploration & Production',
        'Refining & Processing',
        'HSE (Health, Safety, Environment)',
        'Drilling Operations',
        'Pipeline Management'
    ],
    'Information Technology': [
        'Software Development',
        'Cybersecurity',
        'Data Science & AI',
        'Cloud Infrastructure',
        'IT Support',
        'Mobile Development'
    ],
    'Healthcare': [
        'Clinical Operations',
        'Medical Research',
        'Pharmaceutical Development',
        'Hospital Administration',
        'Medical Equipment Maintenance',
        'Pediatric Care'
    ],
    'Banking & Finance': [
        'Retail Banking',
        'Islamic Finance',
        'Investment Management',
        'Risk Analysis',
        'FinTech Innovation',
        'Accounting & Auditing'
    ],
    'Agriculture': [
        'Agronomy Research',
        'Irrigation Technology',
        'Food Processing',
        'Sustainable Farming',
        'Export Operations'
    ],
    'Education': [
        'Higher Education',
        'Vocational Training',
        'Research & Development',
        'International Relations',
        'E-Learning Platforms'
    ]
}

# Job titles from the job dataset
JOB_TITLES = {
    # Energy & Petroleum
    'Exploration & Production': [
        'Petroleum Engineer', 'Reservoir Engineer', 'Geologist', 
        'Geophysicist', 'Production Engineer', 'Field Operations Manager'
    ],
    # Including all departments...
    'Refining & Processing': [
        'Process Engineer', 'Refinery Manager', 'Chemical Engineer', 
        'Quality Control Specialist', 'Operations Supervisor'
    ],
    'HSE (Health, Safety, Environment)': [
        'HSE Manager', 'Safety Inspector', 'Environmental Compliance Officer',
        'Risk Assessment Specialist', 'Safety Training Coordinator'
    ],
    'Drilling Operations': [
        'Drilling Engineer', 'Drilling Supervisor', 'Wellsite Manager',
        'Mud Engineer', 'Directional Driller'
    ],
    'Pipeline Management': [
        'Pipeline Engineer', 'Integrity Manager', 'Pipeline Technician',
        'Project Manager', 'Operations Supervisor'
    ],
    
    # IT Sector
    'Software Development': [
        'Software Engineer', 'Full Stack Developer', 'Backend Developer',
        'Frontend Developer', 'Mobile App Developer', 'DevOps Engineer',
        'QA Engineer', 'Software Architect'
    ],
    'Cybersecurity': [
        'Security Analyst', 'Penetration Tester', 'Security Engineer',
        'Security Consultant', 'CISO', 'Security Operations Specialist'
    ],
    'Data Science & AI': [
        'Data Scientist', 'Machine Learning Engineer', 'Data Analyst',
        'AI Researcher', 'Business Intelligence Analyst', 'Big Data Engineer'
    ],
    'Cloud Infrastructure': [
        'Cloud Architect', 'Cloud Engineer', 'Systems Administrator',
        'Network Engineer', 'Infrastructure Manager'
    ],
    'IT Support': [
        'IT Support Specialist', 'Technical Support Engineer', 'Help Desk Analyst',
        'IT Coordinator', 'Service Desk Manager'
    ],
    'Mobile Development': [
        'Mobile App Developer', 'iOS Developer', 'Android Developer',
        'Mobile UX Designer', 'Mobile QA Tester'
    ],
    # Default/Generic
    'General Management': [
        'Project Manager', 'Department Director', 'Team Leader',
        'Operations Manager', 'General Director', 'Administrative Manager'
    ],
    'Operations': [
        'Operations Manager', 'Production Supervisor', 'Logistics Coordinator',
        'Supply Chain Manager', 'Facilities Manager'
    ]
}

# Skills mapping for reference
SKILLS_MAPPING = {
    # Energy Sector
    'Exploration & Production': [
        'Reservoir Simulation', 'Geological Modeling', 'Well Logging',
        'Enhanced Oil Recovery', 'Petrel Software', 'HYSYS', 'Drilling Optimization'
    ],
    'Software Development': [
        'Python', 'Java', '.NET', 'Django', 'Spring Boot', 'Microservices',
        'REST APIs', 'Git', 'CI/CD Pipelines', 'Agile Methodology'
    ],
    # More skills...
    'Cybersecurity': [
        'Ethical Hacking', 'SIEM Tools', 'Network Security', 'GDPR Compliance',
        'Cryptography', 'Incident Response', 'Firewall Configuration', 'SOC Operations'
    ],
    'Medical Research': [
        'Clinical Trials', 'Biostatistics', 'Epidemiology', 'Laboratory Management',
        'Medical Writing', 'Regulatory Affairs', 'Research Ethics'
    ],
    'Islamic Finance': [
        'Murabaha Transactions', 'Sukuk Structuring', 'Sharia Compliance',
        'Takaful Operations', 'Islamic Banking Products', 'Risk Sharing Models'
    ],
    'Sustainable Farming': [
        'Organic Certification', 'Water Conservation', 'Permaculture Design',
        'Soil Health Management', 'Precision Agriculture', 'Agroecology'
    ]
}

# Additional employee-specific configurations
ALGERIAN_UNIVERSITIES = [
    'University of Algiers', 'University of Science and Technology Houari Boumediene',
    'University of Oran', 'University of Constantine', 'University of Bejaia',
    'University of Tlemcen', 'University of Batna', 'National Polytechnic School of Algiers',
    'University of Annaba', 'University of Setif', 'University of Boumerdes',
    'University of Tizi Ouzou', 'University of Blida', 'University of Biskra',
    'University of Tiaret', 'University of Mostaganem', 'University of Chlef',
    'University of El Oued', 'University of Skikda', 'University of Djelfa'
]

ALGERIAN_LANGUAGES = ['Arabic', 'French', 'English', 'Tamazight', 'Spanish', 'German']

# Function to generate education history
def generate_education_history(age, education_level):
    education_history = []
    
    graduation_age = {
        'Bac': 18,
        'Licence': 21,
        'Master': 23,
        'Doctorat': 28,
        'Ingénieur d\'État': 23
    }
    
    highest_degree_age = graduation_age.get(education_level, 22)
    current_year = datetime.now().year
    graduation_year = current_year - (age - highest_degree_age)
    
    # Add highest degree
    education_history.append({
        'degree': education_level,
        'institution': choice(ALGERIAN_UNIVERSITIES),
        'field_of_study': fake_other.job(),
        'start_year': graduation_year - (3 if education_level in ['Licence', 'Ingénieur d\'État'] else 
                                         2 if education_level == 'Master' else 
                                         5 if education_level == 'Doctorat' else 1),
        'end_year': graduation_year,
        'grade': choice(['Excellent', 'Very Good', 'Good', 'Average'])
    })
    
    # Add previous education if applicable
    if education_level in ['Licence', 'Master', 'Doctorat', 'Ingénieur d\'État']:
        # Add Bac (high school)
        bac_year = graduation_year - (3 if education_level == 'Licence' else 
                                     5 if education_level == 'Master' else 
                                     10 if education_level == 'Doctorat' else 
                                     5 if education_level == 'Ingénieur d\'État' else 0)
        
        education_history.append({
            'degree': 'Bac',
            'institution': fake_other.company() + ' High School',
            'field_study': choice(['Science', 'Mathematics', 'Literature', 'Technical']),
            'start_year': bac_year - 3,
            'end_year': bac_year,
            'grade': choice(['Excellent', 'Very Good', 'Good', 'Average'])
        })
        
    return education_history

# Function to generate work history
def generate_work_history(age, years_experience, sector, department, job_title):
    work_history = []
    current_year = datetime.now().year
    
    # Generate number of previous jobs
    num_prev_jobs = min(randint(0, years_experience // 2), 5)
    
    # Current job details
    current_job_duration = randint(1, years_experience - num_prev_jobs) if years_experience > num_prev_jobs else 1
    current_job_start = current_year - current_job_duration
    
    work_history.append({
        'company': fake_other.company(),
        'job_title': job_title,
        'department': department,
        'sector': sector,
        'start_date': f"{current_job_start}-{randint(1, 12):02d}-{randint(1, 28):02d}",
        'end_date': 'Present',
        'location': choice(ALGERIAN_CITIES),
        'responsibilities': ', '.join(fake_other.bs() for _ in range(3))
    })
    
    # Previous jobs
    remaining_experience = years_experience - current_job_duration
    for i in range(num_prev_jobs):
        if remaining_experience <= 0:
            break
        
        job_duration = min(randint(1, 4), remaining_experience)
        remaining_experience -= job_duration
        end_year = current_job_start - 1
        start_year = end_year - job_duration
        
        prev_sector = choice(ALGERIAN_SECTORS)
        prev_department = choice(DEPARTMENTS.get(prev_sector, ['General Management', 'Operations']))
        
        if prev_department in JOB_TITLES:
            prev_job_title = choice(JOB_TITLES[prev_department])
        else:
            prev_job_title = choice(JOB_TITLES['General Management'])
        
        work_history.append({
            'company': fake_other.company(),
            'job_title': prev_job_title,
            'department': prev_department,
            'sector': prev_sector,
            'start_date': f"{start_year}-{randint(1, 12):02d}-{randint(1, 28):02d}",
            'end_date': f"{end_year}-{randint(1, 12):02d}-{randint(1, 28):02d}",
            'location': choice(ALGERIAN_CITIES),
            'responsibilities': ', '.join(fake_other.bs() for _ in range(2))
        })
        
        current_job_start = start_year
    
    return work_history

# Function to generate employee skills
def generate_employee_skills(department, experience_years):
    base_skills = ['Microsoft Office', 'Project Management', 'Team Leadership', 
                  'Report Writing', 'Arabic/French Bilingual']
    
    # Department-specific skills
    department_skills = SKILLS_MAPPING.get(department, [])
    
    # Generic professional skills
    professional_skills = [
        'Time Management', 'Problem Solving', 'Communication', 'Negotiation',
        'Critical Thinking', 'Presentation Skills', 'Teamwork', 'Adaptability',
        'Conflict Resolution', 'Strategic Planning'
    ]
    
    # Language skills
    language_skills = sample(ALGERIAN_LANGUAGES, k=randint(2, 4))
    language_proficiency = {}
    for lang in language_skills:
        language_proficiency[lang] = choice(['Native', 'Fluent', 'Intermediate', 'Basic'])
    
    # Calculate number of skills based on experience
    num_skills = min(5 + (experience_years // 2), 12)
    
    # Pool all skills together and sample
    all_skills = base_skills + department_skills + professional_skills
    if len(all_skills) > num_skills:
        selected_skills = sample(all_skills, k=num_skills)
    else:
        selected_skills = all_skills
    
    # Return both technical skills and language proficiency
    return {
        'technical_skills': ', '.join(selected_skills),
        'language_proficiency': str(language_proficiency)
    }

# Function to generate salary based on experience and education (LOWERED)
def generate_salary(job_title, years_experience, education_level, seniority_level):
    # Base salary ranges by education level (in Algerian Dinar) - LOWERED VALUES
    base_ranges = {
        'Bac': (25000, 45000),
        'Licence': (35000, 65000),
        'Master': (50000, 85000),
        'Doctorat': (70000, 120000),
        'Ingénieur d\'État': (55000, 90000)
    }
    
    # Multipliers by seniority
    seniority_multipliers = {
        'Junior': 1.0,
        'Mid': 1.2,
        'Senior': 1.4,
        'Lead': 1.6,
        'Chief': 1.8
    }
    
    # Experience bonus (per year) - LOWERED
    experience_bonus = 1500
    
    # Calculate base range
    base_min, base_max = base_ranges.get(education_level, (30000, 60000))
    
    # Apply multipliers
    multiplier = seniority_multipliers.get(seniority_level, 1.0)
    min_salary = base_min * multiplier
    max_salary = base_max * multiplier
    
    # Add experience bonus
    min_salary += years_experience * experience_bonus
    max_salary += years_experience * experience_bonus
    
    # Get random value in range and round to nearest 1000
    salary = randint(int(min_salary), int(max_salary))
    return round(salary / 1000) * 1000

# Function to generate employee dataset
def generate_employee_dataset(num_rows=1000):
    dataset = []
    
    for _ in range(num_rows):
        # Basic demographic information
        first_name = fake.first_name()  # Arabic name only
        gender = choice(['Male', 'Female'])
        age = randint(22, 60)
        birth_date = (datetime.now() - timedelta(days=365*age + randint(0, 364))).strftime('%Y-%m-%d')
        
        # Professional details
        education_level = choice(['Bac', 'Licence', 'Master', 'Doctorat', 'Ingénieur d\'État'])
        seniority_level = choice(['Junior', 'Mid', 'Senior', 'Lead', 'Chief'])
        
        # Experience calculation
        years_experience = max(0, age - 22)  # Assuming education completes by 22 for most
        if education_level == 'Master':
            years_experience = max(0, age - 24)
        elif education_level == 'Doctorat':
            years_experience = max(0, age - 28)
        
        # Sector, department and job title
        sector = choice(ALGERIAN_SECTORS)
        department = choice(DEPARTMENTS.get(sector, ['General Management', 'Operations']))
        
        if department in JOB_TITLES:
            job_title = choice(JOB_TITLES[department])
        else:
            job_title = choice(JOB_TITLES['General Management'])
        
        # Employment details
        employment_type = choice(['Full-Time', 'Part-Time', 'Contract', 'Freelance', 'Internship'])
        contract_type = choice(['CDI', 'CDD', 'Stage', 'Freelance', 'Alternance'])
        
        # Location and contact
        city = choice(ALGERIAN_CITIES)
        address = fake_other.address()
        email = f"{first_name.lower()}@{fake_other.free_email_domain()}"
        phone = "+213" + ''.join([str(randint(0, 9)) for _ in range(9)])
        
        # Generate detailed backgrounds
        education_history = generate_education_history(age, education_level)
        work_history = generate_work_history(age, years_experience, sector, department, job_title)
        skills_data = generate_employee_skills(department, years_experience)
        
        # Generate salary (lowered)
        current_salary = generate_salary(job_title, years_experience, education_level, seniority_level)
        
        # Job seeking status
        job_seeking_status = choice(['Not looking', 'Open to offers', 'Actively looking', 'Recently employed'])
        
        # Social media profiles (randomly include or exclude some)
        social_media = {}
        if random() > 0.3:  # 70% have LinkedIn
            social_media['linkedin'] = f"linkedin.com/in/{first_name.lower()}-{randint(100, 999)}"
        if random() > 0.5:  # 50% have Facebook
            social_media['facebook'] = f"facebook.com/{first_name.lower()}.{randint(10, 99)}"
        if random() > 0.7:  # 30% have Twitter
            social_media['twitter'] = f"twitter.com/{first_name.lower()}{randint(1, 999)}"
        
        # Create the record
        record = {
            # Personal Information
            'first_name': first_name,  # Arabic name only
            'gender': gender,
            'birth_date': birth_date,
            'age': age,
            'email': email,
            'phone': phone,
            'address': address,
            'city': city,
            
            # Professional Summary
            'current_job_title': job_title,
            'department': department,
            'sector': sector,
            'seniority_level': seniority_level,
            'years_experience': years_experience,
            'highest_education': education_level,
            'employment_type': employment_type,
            'contract_type': contract_type,
            
            # Skills
            'technical_skills': skills_data['technical_skills'],
            'language_proficiency': str(skills_data['language_proficiency']),
            
            # Financial
            'current_salary_dzd': current_salary,
            
            # Status
            'job_seeking_status': job_seeking_status,
            'availability_date': (datetime.now() + timedelta(days=randint(0, 90))).strftime('%Y-%m-%d') if job_seeking_status in ['Open to offers', 'Actively looking'] else None,
            
            # Social Media
            'social_media': str(social_media),
            
            # Detailed Histories
            'education_history': str(education_history),
            'work_history': str(work_history)
        }
        
        dataset.append(record)
    
    return pd.DataFrame(dataset)

# Function to generate job dataset
def generate_job_dataset(num_rows=1000):
    dataset = []
    
    for _ in range(num_rows):
        sector = choice(ALGERIAN_SECTORS)
        department = choice(DEPARTMENTS.get(sector, ['General Management', 'Operations']))
        
        # Fixed job title selection
        if department in JOB_TITLES:
            job_title = choice(JOB_TITLES[department])
        else:
            job_title = choice(JOB_TITLES['General Management'])
        
        post_date = fake_other.date_between(start_date='-30d', end_date='today')
        deadline = post_date + timedelta(days=randint(7, 60))
        
        raw_salary = randint(25000, 150000)  # LOWERED salary range
        
        record = {
            'job title': job_title,
            'hiring organization': choice(ALGERIAN_COMPANIES),
            'department': department,
            'seniority level': choice(['Junior', 'Mid', 'Senior', 'Lead', 'Chief']),
            'date posted': post_date.strftime('%Y-%m-%d'),
            'application deadline': deadline.strftime('%Y-%m-%d'),
            'valid through': (deadline + timedelta(days=30)).strftime('%Y-%m-%d'),
            'sector': sector,
            'job location': choice(ALGERIAN_CITIES),
            'education requirement': choice(['Bac', 'Licence', 'Master', 'Doctorat', 'Ingénieur d\'État']),
            'education_years': randint(12, 20),
            'type of contract': choice(['CDI', 'CDD', 'Stage', 'Freelance', 'Alternance']),
            'available posts': randint(1, 15),
            'description': fake_other.paragraph(nb_sentences=5),
            'benefits': ', '.join(sample([
                'Assurance maladie', 'Voiture de fonction', 
                'Tickets restaurant', 'Télétravail partiel',
                'Formation continue', 'Prime de performance',
                'Épargne entreprise', 'Crèche d\'entreprise'
            ], 3)),
            'predicted_salary_dzd': round_salary(raw_salary),
            'inferred_skills': generate_skills(department),
            'experience_min_req': randint(0, 5),
            'experience_max_req': randint(3, 15)
        }
        dataset.append(record)
    
    return pd.DataFrame(dataset)

# Helper function for rounding salary
def round_salary(amount):
    return round(amount / 1000) * 1000

# Helper function to generate skills
def generate_skills(department):
    base_skills = ['Microsoft Office', 'Project Management', 'Team Leadership', 
                  'Report Writing', 'Arabic/French Bilingual']
    
    # Get department-specific skills if available
    department_skills = SKILLS_MAPPING.get(department, base_skills)
    
    # Make sure we don't try to sample more items than are in the list
    num_skills = min(len(department_skills), 8)
    if num_skills > 0:
        skills = sample(department_skills, num_skills)
    else:
        skills = base_skills
        
    return ', '.join(skills)

# Generate and save datasets
employee_df = generate_employee_dataset(8124)
employee_df.to_csv('algerian_employees_dataset.csv', index=False)
print("Algerian employees dataset generated with shape:", employee_df.shape)

job_df = generate_job_dataset(1248)
job_df.to_csv('algerian_jobs_dataset.csv', index=False)
print("Algerian jobs dataset generated with shape:", job_df.shape)

# Display samples
print("\n--- Employee Sample ---")
print(employee_df.sample(3).to_string())
print("\n--- Job Sample ---")
print(job_df.sample(3).to_string())
```

```python
import csv

def remove_columns_from_csv(input_file, output_file, columns_to_remove):
    # Open the input and output files with UTF-8 encoding
    with open(input_file, 'r', encoding='utf-8', newline='') as infile, \
         open(output_file, 'w', encoding='utf-8', newline='') as outfile:
        
        reader = csv.reader(infile)
        writer = csv.writer(outfile)
        
        # Read the header
        header = next(reader)
        
        # Find indices of columns to remove
        column_indices = []
        for column in columns_to_remove:
            if column in header:
                column_indices.append(header.index(column))
            else:
                print(f"Column '{column}' not found in the CSV file.")
        
        # Sort indices in descending order to avoid shifting
        column_indices.sort(reverse=True)
        
        # Create new header
        new_header = header.copy()
        for index in column_indices:
            del new_header[index]
        
        # Write the new header
        writer.writerow(new_header)
        
        # Process and write each row
        for row in reader:
            new_row = row.copy()
            for index in column_indices:
                if index < len(new_row):
                    del new_row[index]
            writer.writerow(new_row)
        
    print(f"Columns {columns_to_remove} removed successfully!")

# === Example Usage ===
input_file = 'algerian_jobs_dataset.csv'  # Change this if needed
output_file = 'jobs_modified.csv'             # Output file name
columns_to_remove = [
    'application deadline' , 'valid through' , 'education_years'
]

remove_columns_from_csv(input_file, output_file, columns_to_remove)
```
